## 야근(OverTime)을 하는 직원과 야근을 하지 않는 직원의 이직률 차이가 얼마나 나는지 비교하고 결과를 서술하세요.
- 범주형 vs 범주형 데이터

```python
# .crosstab()
# normalize= : 값을 백분율로 만들어줌
overtime_attrition = (
    pd.crosstab(hr_df["OverTime"],hr_df["Attrition"], normalize="index")* 100
)

print("야근 여부에 따른 이직 비율")
overtime_attrition
```

```python
overtime_rate = hr_df.groupby("OverTime")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100)
overtime_rate
```

```python
ax = overtime_attrition.plot(kind="bar", stacked=True)
plt.title("야근 여부에 따른 이직 비율(%)")
plt.xlabel("야근 여부")
plt.xticks(rotation=0)
plt.ylabel("비율(%)")
plt.show()
```

## 검증
- 야근을 하는 직원의 이직률이 야근을 하지 않는 지원의 이직률보다 약 20% 더 높다


- 범주형끼리 데이터를 비교할 떄 Bar 차트가 적합하다

## 3번
- 마지막 승진 후 기간(YearsSinceLastPromotion)이 길어질수록 이직률이 더 높아질까?
- 검증: 데이터의 연속형, 순서형 데이터들을 시각화 -> 라인차트를 권장

```python
at_year_df = hr_df.groupby("YearsSinceLastPromotion")["Attrition"].agg(
    YesRate = lambda x: (x == "Yes").mean() * 100,
    NoRate = lambda x: (x == "No").mean() * 100
).reset_index()
at_year_df
```

```python
plt.figure(figsize=(10, 5))
plt.plot(at_year_df["YearsSinceLastPromotion"], at_year_df["YesRate"], linewidth=2.5, label="이직")
plt.plot(at_year_df["YearsSinceLastPromotion"], at_year_df["NoRate"], linewidth=2.5, label="유지")
plt.title('마지막 승진 후 기간별 이직률 추이')
plt.xlabel('마지막 승진 후 기간 (년)')
plt.ylabel('이직률')
plt.axvline(3, alpha=0.3, linestyle="--")
plt.axvline(6, alpha=0.3, linestyle="--")
plt.axvline(9, alpha=0.3, linestyle="--")
plt.legend()
plt.show()

"""
    승진 직후 0년차에 약 18.9% 정도의 이직률을 갖는다.
    승진 경력의 타이틀을 호가보한 직원이 몸값을 올려서 이직하지 않을까?
    1~2년 승진을 못하면 승진을 포기하고 이직을 하는 비율도 생각보다 높게 나타났다.
    장기화 7년차(21%) 이후부터는 다시 이직률(0%)이 낮아지는 역전 현상이 일어나고 있다 - 데이터부족
    승진 후 기간이 길어질수록 이직률이 높아진다는 부분 기각
"""
```
