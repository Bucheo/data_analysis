# shape
- numpy에서는 해당 array의 크기를 알 수 있다. shape를 확인함으로서, 몇 개의 데이터가 있는지 몇 차원으로 존재하는지 확인할 수 있다.
- arr.shape의 결과는 튜플로 (행, 열)로 행만 존재하면 1차원, 행렬이 존재하면 2차원으로 차수를 알 수 있다. 또한 행*렬을 곱하면 크기도 확인할 수 있다.

## np.reshape()

## 차원축(axis)
- 1차원 array라면 각 차원에 대한 axis = 0
- 2차원 array라면 각 차원에 대한 axis = 0, 1
- 3차원 array라면 각 차원에 대한 axis = 0, 1, 2

```python
arr1 = np.arange(15)
arr1
```

```python
np.sum(arr1)
```

```python
arr1 = np.arange(15).reshape(3, 5)
arr1
```

```python
print(np.sum(arr1, axis=0))
print(np.sum(arr1, axis=1))
```

```python
arr2 = np.arange(24).reshape(2, 3, 4)
arr2
```

```python
print(np.sum(arr2, axis=0), end="\n\n")
print(np.sum(arr2, axis=1), end="\n\n")
print(np.sum(arr2, axis=2), end="\n\n")
print(np.sum(arr2, axis=-1), end="\n\n")
```

## ★★★argmax() 자주쓰는 문법
- 최대값의 인덱스

```python
arr3 = np.arange(5, 10)
print(arr3)
print(arr3.argmax())
print(arr3[arr3.argmax()])
```

```python
np.random.seed(918)

arr4 = np.random.randint(10, size=(1, 6)).flatten()
print(arr4)
print(arr4.argmax())
print(arr4.argmin())
print(arr4.max())
print(arr4.min())
```

```python
arr5 = np.random.randint(10, size=(3, 6))
print(arr5, end="\n\n")
print(arr5.argmax(axis=0), end="\n\n")
print(arr5.argmax(axis=1), end="\n\n")
print(arr5.max(axis=0), end="\n\n")
print(arr5.min(axis=1), end="\n\n")
```

## boolean selection, boolean indexing
- array 인덱싱 시, bool 타입의 조건에 맞는 list를 전달하여 조건식을 만족하는 True의 결과물만 필터링한다.

```python
arr4 =  np.random.randint(1, 101, size=20)
print(arr4)
```

```python
even_mask = arr4 % 2 == 0
arr4[even_mask]
```

## 다중 조건식
- & : and 조건식
- | : or 조건식

```python
# 홀수이면서 50보다 작은 값
# and 조건 필요
np.random.seed(2026)

arr5 = np.random.randint(100, size=15)
print(arr5)

is_bound = (arr5 % 2 != 0) & (arr5 < 50)
print(arr5[is_bound])

print(arr5[is_bound].astype(str))
```
