```python
# 1등석과 3등석의 생존율 비교
survived_mask = tatanic_df["Survived"] == 1
sur = len(tatanic_df[survived_mask])

pclass_first_mask = tatanic_df["Pclass"] == 1
first = len(tatanic_df[pclass_first_mask & survived_mask])

pclass_second_mask = tatanic_df["Pclass"] == 2
second = len(tatanic_df[pclass_second_mask & survived_mask])

pclass_third_mask = tatanic_df["Pclass"] == 3
third = len(tatanic_df[pclass_third_mask & survived_mask])

print(len(tatanic_df[survived_mask]))
print(f"1등석 생존자 수: {first}, 생존률: {round(first / sur, 2) * 100}%")
print(f"1등석 생존자 수: {second}, 생존률: {round(second / sur, 2) * 100}%")
print(f"3등석 생존자 수: {third}, 생존률: {round(third / sur, 2) * 100}%")
```

```python
pclass_1_mask = tatanic_df["Pclass"] == 1
pclass_3_mask = tatanic_df["Pclass"] == 3

pclass_1_survived = len(tatanic_df[survived_mask & pclass_1_mask])
pclass_3_survived = len(tatanic_df[survived_mask & pclass_3_mask])

print(f"타이타닉호의 1등석 생존률: {pclass_1_survived / len(tatanic_df[pclass_1_mask]) * 100:.2f}%")
print(f"타이타닉호의 3등석 생존률: {pclass_3_survived / len(tatanic_df[pclass_3_mask]) * 100:.2f}%")
```

## 파생 변수 만들기 연습

```python
# 만약 NaN이라면: 알수없음
# 19세 이상: 성인
# 19세 미만: 아이
# if ~ elif ~ else
def get_adult(age):
    if np.isnan(age):
        return "알 수 없음"

    return "성인" if age >= 19 else "아이"

get_adult(18)
```

```python
# apply(): 앞의 시리즈의 값을 순차적으로 받아서 결과를 바꾸고 새로운 결과를 반환하는 함수
tatanic_df["IsAdult"] = tatanic_df["Age"].apply(get_adult)
tatanic_df
```

```python
# RipOff
# Fare > 300: 바가지 요금
# Fare <= 300: 정상 요금

tatanic_df["RipOff"] = tatanic_df["Fare"].apply(lambda fare: "흑우" if fare > 300 else "정상")
tatanic_df[tatanic_df["Fare"] > 300]
```
