## Series
   - Pandas의 기본 객체 중 하나로, numpy의 ndarray를 기반으로 인덱싱 기능이 추가된
   1차원 배열을 나타낸다. (index x value의 형태)
   - index를 지정하지 않을 시, 기본적으로 ndarray와 같이 0-based 인덱스 생성한다
   - index를 지정할 경우 명시적으로 지정된 index를 사용한다.
   - 같은 타입의 0개 이상의 데이터를 가질 수 있다

## Series 함수
    - ndim : 차원 반환
    - len : 데이터 길이
    - size : 개수 반환
    - shape : 튜플의 형태로 shape반환
    - unique : 유일한 값만 udarray로 반환
    - count : NaN을 제외한 개수를 반환
    - mean : NaN를 제외한 평균
    - value_counts : NaN를 제외하고 각 값들의 빈도를 반환

## NaN(Not a Number)
- 숫자형 데이터가 들어와야 하는데 ㅜㅅ자형 데이터가 아닌 다른 값이 들어왔을 때 나타나는 값
- ex) DBMS -> null, JS -> undefined, python -> None

```python
s1 = pd.Series([10, 2, 3, 3, np.nan, 2, 3, 5, 7, 8, np.nan, np.nan])
s1
```

```python
print(s1.size)
print(s1.shape)
print(len(s1))
```

```python
print(s1.unique()) # 
print(s1.count()) # NaN값을 제외하고 실제 값이 들어있는 값이 몇개인지 알려줌
print(s1.value_counts()) # 값마다 총 몇 개가 있는지 알려줌
```

```python
# 디폴드가 위에서부터 5개 값을 보여줌
print(s1.head(), end="\n\n")
print(s1.head(3), end="\n\n")

# 디폴드가 뒤에서부터 5개 값을 보여줌
print(s1.tail(), end="\n\n")
print(s1.tail(3))
```

## 시리즈(Series)의 연산
- 시리즈는 시리즈끼리 연산이 가능하다. 단, index를 기준으로 연산된다.
- 이 특징은 DataFrame에서도 동일하게 적용된다.

```python
products = {
    "치킨" : 30000,
    "요거트": 50000,
    "그릭 요거트": 13000,
    "커피": 6000,
    "떡볶이": 20000,
    "샌드위치": 50000
}
s2 = pd.Series(products)
s2
```

```python
products2 = {
    "치킨" : 30000,
    "요거트": 50000,
    "음료수": 5000,
    "라면": 6000,
    "두쫀쿠": 10000,
    "피자": 20000
}
s3 = pd.Series(products2)
s3
```

```python
s2 * s3
```

```python
# 3만원이 넘는 음식 가지고 오기
price_mask = s2 > 30000
print(price_mask)
print(s2[price_mask])
```

```python
hobbies = {
    "승현": "칵테일방, 찜질방, 노래방",
    "예나": "요리, 노래방, 영화",
    "진성": "호프, 노래방, 술집",
    "가나디": "그림, 그릭요거트, 영화",
    "지혁": "음주, 흡연, 도박",
    "시우": "볼링, 클라이밍, 보석십자수"
}

s4 = pd.Series(hobbies)
```

```python
# 값에 호프, 도박, 흡연란 단어가 있으면 제거기기
evil_list = ["호프", "도박", "흡연"]
evil_mask = [sum(evil in hobby for evil in evil_list) == 0 for hobby in s4]

evil_member = s4[evil_mask]
print(evil_member)
```

```python
# contains 각행을 접근해서 글자들이 포함되어 있으면 true, 없으면 false
print(s4.str.contains("호프|도박|흡연"))
print(s4[s4.str.contains("호프|도박|흡연") == 0])
```

## groupby()

```python
salary_list = [
    "지혁", "희수", "승현", "우영", "희수", "시우", "슬기", "상아", "상아", "희수", "우영", "진휘", 
    "지혁", "준승", "예나", "진휘", "지호", "지혁", "희수", "승현", "시우", "준승", "우영", "예나"
]

salary_income = [
    200, 300, 400, 400, 350, 500, 300, 600, 600, 500, 600, 500,
    300, 300, 500, 500, 300, 400, 300, 800, 300, 200, 700, 500
]
```

```python
s1 = pd.Series(data=salary_income, index=salary_list)
s1
```

```python
print(s1.groupby(s1.index).count(), end="\n\n")
print(s1.groupby(s1.index).sum(), end="\n\n")
print(s1.groupby(s1.index).mean(), end="\n\n") # 평균
```

```python
s1.groupby(s1 >= 300).count()
```

```python
# 600만원 이상 받은 사용자의 수
print(s1.groupby(s1 >= 600).count())
print(s1.groupby(s1 >= 600).apply(lambda x: (x.index, x)))
print(s1.groupby(s1 >= 600).apply(lambda x: list(zip(x.index, x))))
```

```python
# to~: 형변환
# is~: bool
# get~: getter 값 반환
# set~: setter 값 반환

print(s1.to_dict())
print(s1.to_list())
```
