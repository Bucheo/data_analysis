```python
my_python = "Python 3.12 & AI Agent! Python is Powerful, Python is Easy."
word_dict = {}

my_python = my_python.lower() 

for char in my_python: 
    word_dict[char] = word_dict.get(char, 0) + 1  

for key, value in word_dict.items():
    print(f"{key}: {value}개")
```

```python
# 5. 마켓 문제 (40점)
 
# 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
# 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
# - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
# - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
# - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.
# ex) 유저의 할인율이 10%라면 2700원이 결제된다.
# - 판매된 상품의 모든 수익금은 income에 누적된다.

# 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(money), 할인율(discount)의 필드를 갖는다.
# 2) 총 3명의 Buyer를 각각 객체화
# - 홍길동, 15000, 5
# - 장보고, 10000, 20
# - 이순신, 30000, 50
 

# 구매자의 할인율을 적용받아 마켓에서 토마토를 구매를 아래와 같이 구매하였다.
# 마켓의 총 수익과 각 구매자의 잔액을 출력하시오.
# - 홍길동 6개 구매
# - 장보고 2개 구매
# - 이순신 8개 구매
```

```python
class Market:
    #생성자
    def __init__(self, name, product, income = 0):
        self.name = name
        self.product = product
        self.income = income

    #판매
    def sell(self, buyer, count):
        tomato_price = self.product["price"] # 토마토 가격
        tomato_discount_price = int(tomato_price * ((100 - buyer.discount) / 100)) # 사용자 마다의 할인된 가격
        product_stock = self.product["stock"] # 토마토 재고
        total_price = tomato_discount_price * count # 결제 총 금액
        name = buyer.name # 사용자의 이름
        
        if buyer.money < total_price:
            print(f"""
                {name}님 돈이 부족하여 구매가 불가합니다.
                총 구매가격: {total_price}, 구매자의 현재 돈: {buyer.money}
                """)
            return
        elif count > product_stock:
            print(f"""{name}님 재고이 부족하여 구매가 불가합니다.
            현재 재고: {product_stock}, 구매자의 구매 갯수: {count}
            """)
            return

        self.income += (tomato_discount_price * count)
        self.product["stock"] -= count

        print(f"""
            {name}님 구매해주셔서 감사합니다!
            (할인된 가격: {tomato_discount_price}, 구매 갯수: {count}, 총 금액: {tomato_discount_price * count})
            재고: {self.product["stock"]}\t 총 수익금: {self.income}원
            """)

    #수입금 출력
    def show_income(self):
        print(f"현재 총 수익금은 {self.income}원 입니다.")
    
class Buyer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount

#객체화
market = Market("이마트", {"name":"토마토", "price":3000, "stock":10})
hong = Buyer("홍길동", 15000, 5)
jang = Buyer("장보고", 10000, 20)
lee = Buyer("이순신", 30000, 50)

market.sell(hong, 6)
market.sell(jang, 2)
market.sell(lee, 8)
```
