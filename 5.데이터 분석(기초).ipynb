## 계급(Class)
- 변량을 일정 간격으로 나눈 구간을 의미한다.
- 변량의 최솟값과 최댓값을 잘 고려해서 계급을 정해야한다.

```python
df_seoul = df["서울"]
df_seoul
```

```python
df_class = pd.cut(df_seoul, bins=[140, 150, 160, 170, 180, 190])
df_class
```

```python
df_seoul_class = pd.DataFrame({"서울": df_seoul, "계급":df_class})
df_seoul_class
```

## 도수
- 각 계급에 속하는 변량의 개수를 의미한다.

```python
df_seoul_class.groupby("계급", observed=True).count().apply(lambda x: x / 100)
```

## 도수 분포표
- 주어진 자료를 계급별로 나눈 뒤 각 계급에 속하는 도수 및 상대 도수를 조사한 표이다.
- 구간별 분포를 한 번에 알아보기 좋지만 계급별 각 변량의 정확한 값이 생갹되었다.

```python
freqnency_df = pd.DataFrame({
    "도수": df_seoul_class.groupby("계급", observed=True).count()["서울"],
    "상대 도수": df_seoul_class.groupby("계급", observed=True).count().apply(lambda x: x / 100)["서울"]
})
freqnency_df
```

## 산술 평균(Mean)
- 변량의 합을 변량의 수로 나눈 값을 의미한다
![](https://drive.google.com/thumbnail?id=1NDSN0HsfG0uksCUPO17O4eTSGuSC3jOJ&sz=w800)

```python
df.mean().to_frame(name="지역별 평균 키")
```

## 편차(Deviation)

```python
gyeonggi_df = df["경기"].to_frame()
gyeonggi_df["편차"] = gyeonggi_df["경기"].apply(lambda x: x - gyeonggi_df.mean())
gyeonggi_df
```

## 분산(Variance)
- 이상치를 구하기 위해 평균으로 부터 얼마나 떨어졌는지 알아내기 위한 값

1. 절대 값으로 구한다
2. 모든 편차의 제곱을 하고 그 합의 평균을 낸다.

```python
gyeonggi_df["편차의 제곱"] = gyeonggi_df["편차"].apply(lambda x: x ** 2)
```

```python
variance = gyeonggi_df["편차의 제곱"].mean()
variance
```

## 표준 편차(Standard deviation)
- 분산의 제곱근이며, 관측된 변량의 흩어진 정로를 하나의 수치로 나타내는 통계량이다.
- 표준편차가 작을 수록 평균값에서 변량들의 거리가 가깝다고 판단한다.

```python
import math

# sqrt: 스퀘어 루트
std = math.sqrt(variance)
std
```

### 확률 변수(Random variable)
- 머신러닝, 딥러닝 등 확률을 다루는 분야에 있어서 필수적인 개념이다.
- 확률(probability)이 있다는 뜻은 사건(event)이 있다는 뜻이며,
  시행(trial)을 해야 시행의 결과인 사건(event)이 나타난다.
- 시행을 해야 어떤 사건이 나타났는지에 따라 값이 정해지는 변수(함수)이다.
- 알파벳 대문자로 표현하며, X, Y, Z와 같이 표현한다.
- 확률 변수의 실현값들의 집합을 상태 공간이라 하며, 원소를 확률변수값(실현값)이라고 표현한다.
- 확률 변수에서 사용한 알파벳의 소문자를 사용한다.
- S(표본공간) = {y<sub>1</sub>, y<sub>2</sub>, y<sub>3</sub>}
- 이 때 Y는 확률 변수이고, 원소인 y<sub>1</sub>, y<sub>2</sub>, y<sub>3</sub>은 실현값이다.

#### 범주형 확률 변수(Categorical random variable)
- 범주형 확률 변수 값은 수치가 아닌 기호나 언어, 숫자등으로 표현하고, 기호나 언어는 순서를 가질 수 있다.
- 유한 집합으로 표현한다. 원소의 수가 유한한 집합을 의미한다.
- {봄, 여름, 가을, 겨울}, {앞면, 뒷면}, {선택, 미선택}

#### 이산형 확률 변수(Discrete random variable)
- 이산형 확률 변수 값은 수치로 표현하고 셀 수 있는 값이다.
- 이를 더 넓은 범위로 양적 확률 변수 또는 수치형 확률 변수라고도 한다.
- 유한집합 또는 셀 수 있는 무한 집합으로 표현한다. 무한 집합은 원소의 수가 무한한 집합을 의미한다.
- {0, 1, 2, 3}, {10, 20, 30, ...}

#### 연속형 확률 변수(Continuous random variable)
- 연속형 확률변수는 구간을 나타내는 수치로 표현한다. 이를 더 넓은 범위로
  양적 확률 변수 또는 수치형 확률 변수라고도 부른다
- 셀 수 없는 무한 집합으로 표현한다. 셀 수 없으므로 범위로 표현한다
- 128.56 < x < 268.56

## 확률 분포(Probability distribution)
- 사건에 대한 확률 변수에서 정의된 모든 확률값의 분포이며, 서로 다른 모든 결과의 출현 결과를 제공한다.

1. 동전 던지기  -  시행
2. {0, 1}  -  확률 변수와 확률 변수 값
3. 완벽한 형태의 동전일 때 확률 분포


![](https://drive.google.com/thumbnail?id=1MzWxGS0PvIjOHTw8yZSe-e_E9xUKirWH&sz=w800)

## 자연의 법칙, "중심극한정리"
이게 동전뿐만 아니라 사람의 키, 시험 점수, 심지어 전철 연착 시간까지 독립적인 수많은 우연이 겹치는 자연계의 모든 현상에서 똑같이 일어난다. 이 우연들이 무한히 모여서 만들어진 부드러운 종 모양의 곡선을 가우시안 분포(정규분포)라고 부른다.
동전을 10번이 아니라 100번, 1,000번, 10,000번으로 늘리고 기둥의 폭을 아주 촘촘하게 만들면, 결국 각진 기둥들이 부드러운 하나의 곡선(종 모양)으로 변합니다. 이 곡선이 바로 우리가 아는 정규분포(Normal Distribution)이자, 머신러닝에서 말하는 가우시안 분포(Gaussian Distribution)입니다. 즉, 동전 던지기 같은 이산형 확률 분포(이항분포)도 시행 횟수를 엄청나게 늘리면 결국 연속형 확률 분포인 '정규분포'가 됩니다.

## 확률 분포표
- 확률 변수의 모든 값(원소)에 대해 확률을 표로 표시한 것

```python
df_h_dist = pd.DataFrame(np.arange(0, 100) % 4 + 1, columns=["경주마 번호"])

df_h_hist_group = df_h_dist.groupby("경주마 번호")["경주마 번호"].count().reset_index(name="1등 횟수")

df_h_hist_group["1등 확률"] = df_h_hist_group["1등 횟수"] / 100
df_h_hist_group
```

```python
plt.hist(df_h_dist["경주마 번호"], range(1, 6), density=True, color="pink", edgecolor="orange")
plt.xlim(0, 6)
plt.ylim(0, 0.3)
plt.show()
```

### 확률 질량 함수(Probability mass function, pfm)
- 확률 변수 X의 분포를 나타내는 함수로서, x<sub>i</sub>가 나올 확률이다.
- 확률 변수의 값을 입력 받고, 해당 값이 나타날 확률을 구해서 출력하는 함수이다.
- 범주형 확률 변수와 이산형 확률 변수에서 사용된다
- 확률 변수에서 각 값에 대한 확률을 나타내는 것이 마치 각 값이 "질량"을 가지고 있는 것처럼 보이기 때문에 확률 질량 함수로 불린다.

> 확률 질량 함수 f는 확률 변수 X가 x를 변수값으로 가질 때 확률이다.

![](https://drive.google.com/thumbnail?id=1du4Y69yvH4g1FcNzYDTlDtKt_z7Y6IHE&sz=w800)

- 쉽게 말해 확률의 너비를 구하면 질량(확률)을 구할 수 있다. 확률의 합은 1이므로 각각의 확률을 땅따먹기 처럼 나눈 것과 같다.

## 확률 밀도 함수(Probability Density Function, pdf)
- 확률변수 X의 분포를 나타내는 함수로서, 특정 구간에 속할 확률이고 이는 특정 구간을 적분한 값이다.
- 확률 변수값의 범위(구간)을 입력 받고, 범위의 넓이를 구해서 리턴하는 함수이다
- 연속형 확률변수에서 사용된다
- 전체에 대한 확률이 아닌 구간에 포함될 확률을 나타내기 때문에 구간에 따른 밀도를 구하는 것이고, 이를 확률 밀도 함수라 불린다.

### 적분(Integral): $\int$
- 선분 = 높이(길이)
- 면적을 구할 때 여러 사각형으로 나눈 뒤 합하여도 전체 면적이 나온다.
- 가로가 무한소 0인 사각형을 쌓은 뒤 각 면적을 모두 합하는 것이 적분이다.

- 둥근 원이 있을 때 작은 나무토박을 촘촘하게 둬서 공간을 구하는 것
- 확률 밀도 함수를 구해야하기 때문


![](https://drive.google.com/thumbnail?id=1-Y2BBBRDCDrFn6nzPlEQ503lJkh5QYSd&sz=w800)


> $$P(a < X < b) = \int_{a}^{b} f(x) dx$$
> 우변의 $\int_{a}^{b} f(x) dx$은 곡선 $f(x)$ 아래에서 $a$부터 $b$까지의 면적을 구하라
